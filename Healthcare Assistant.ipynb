#### **Download LLaMA-2 13B Chat Model**

We are going to download and use the Llama model which is trained on 13 billion parameters. The size of this model is around 9GB so it is recommended to have a good internet connection along with a GPU to download it and generate responses from it respectively.

````python
model_name_or_path = "TheBloke/Llama-2-13B-chat-GGUF"
model_basename = "llama-2-13b-chat.Q5_K_M.gguf"
model_path = hf_hub_download(                     # Download the model file from Hugging Face Hub and return its local path
    repo_id=model_name_or_path,
    filename=model_basename
    )
````

**Model path :** Specifies the Hugging Face repository containing the Llama-2-13B GGUF model.

**Model Name :** Specifies the exact quantized GGUF model file to download and use for inference.

**Model Download :** Downloads the selected Llama-2 model from Hugging Face Hub and returns the local file path for loading the model.

#### **Initialize LLaMA Model with Configuration**

````python
# The below snippet of code if the runtime is connected to GPU.
lcpp_llm = Llama(
     model_path = model_path, # Path to the downloaded GGUF model
     n_cpu_threads=4, # Number of CPU threads to use
     n_batch=512, # Batch size for prompt processing
     n_gpu_layers=40, # Number of layers to offload to GPU
     n_ctx=4096 # Context window
)
````

The given code initializes a **Llama model**for local inference. Here's a breakdown of each parameter:

- **`Llama(...)`**: Loads a **Llama model** for text generation.
- **`model_path=model_path`**: Specifies the **file path** of the downloaded model (from Hugging Face or another source).
- **`n_ctx=..`**: Sets the **context window** (i.e., the maximum number of tokens the model can process at once).
- **`n_batch=..`**: Defines the **batch size** for processing tokens. A higher value improves speed but requires more VRAM.
- **`n_gpu_layers=..`**: Determines how many **layers** are offloaded to the **GPU**. Adjust this based on available **VRAM**.

This setup is optimized for **running a local Llama model**, leveraging both **CPU and GPU** for efficient inference. The parameters should be adjusted based on **hardware constraints** (CPU, GPU, and RAM availability).

````python
# Provides an example and answer anchor to guide the model in giving concise, evidence-based responses without echoing the question.
system_prompt = """
You are a medical expert AI assistant. Respond with concise, evidence-based answers.
"""

user_prompt = """
Example:
Q: What are the common symptoms of appendicitis?
A: Common symptoms include abdominal pain (usually starting near the navel), nausea, vomiting, and fever.
References: Mayo Clinic, UpToDate

Now answer:
Q: What is the protocol for managing sepsis in a critical care unit?
A:
"""
````

#### **Response Function**

````python
# function to generate, process, and return the response from the LLM
def prompt_engineering_response(user_prompt):
    # Put the system message first, then the user question, then anchor with "Answer:"
    prompt = f"""{system_prompt}

Question: {user_prompt}

Answer:"""

    # Generate a response from the LLaMA model
    response = lcpp_llm(
        prompt=user_prompt,
        max_tokens=512,
        temperature=0.2,
        top_p=0.95,
        repeat_penalty=1.2,
        top_k=25,
        stop=['INST'],
        echo=False
    )

    # Extract and return the response text
    response_text = response["choices"][0]["text"].strip()
    return response_text
````

## **Retriever**

Retrieval and Response Generation using Vector Search

````python
retriever = vectorstore.as_retriever(
    search_type='similarity',
    search_kwargs={'k': 3}
)
````

````python
medical_system_message = """
You are an AI assistant designed to support healthcare professionals by providing evidence-based, concise, and accurate responses using authoritative medical sources, such as the Merck Manuals.

Your goal is to help clinicians, researchers, and healthcare teams quickly access reliable medical knowledge to improve patient outcomes, support decision-making, and reduce information overload.

User input will include context extracted from trusted medical sources. This context will begin with the token:

###Context
The context may include excerpts from the Merck Manuals, clinical guidelines, or peer-reviewed medical literature, including titles, sections, authors, and other relevant metadata.

When crafting your response:
- Use only the provided context to answer the question.
- Provide concise, clinically relevant, and accurate answers.
- Include the source (title, section, and page/section reference) when applicable.
- If the context does not contain relevant information, respond: "Sorry, this is out of my knowledge base."
- Do NOT provide personal medical advice or treatment recommendations outside of the context.
- Maintain a professional, neutral, and safe tone appropriate for healthcare communication.

Example response format:

Answer:
[Answer based on context]

Source:
[Source title, section, page]
"""
````

````python
medical_user_message_template = """
###Context
Here are relevant excerpts from the Merck Manuals or other authoritative medical sources:
{context}

###Question
{question}
"""
````

### Response Function

````python
def generate_rag_response(user_input, retriever,
                          system_message, user_message_template,
                          k=5, max_tokens=500, temperature=0.3, top_p=0.95):

    # Retrieve relevant document chunks
    relevant_chunks = retriever.invoke(user_input)
    if not relevant_chunks:
        return "Sorry, this is out of my knowledge base."

    # Combine document chunks with source info
    context_for_query = "\n".join([f"Source: {d.metadata.get('source','Unknown')}\n{d.page_content}" for d in relevant_chunks])

    # Fill user message template to get the user's part of the prompt
    user_message_content = user_message_template.format(context=context_for_query, question=user_input)

    # Construct the full prompt for the Llama model
    full_prompt = f"""{system_message}

{user_message_content}

Answer:"""

    # Generate the response using the globally available lcpp_llm object
    try:
        response = lcpp_llm(
            prompt=full_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            repeat_penalty=1.2,
            top_k=25,
            stop=['INST'],
            echo=False
        )
        return response["choices"][0]["text"].strip()

    except Exception as e:
        return f"Sorry, I encountered the following error:\n{e}"
````

# **Output Evaluation**

````python
medical_groundedness_rater_system_message = """
You are a strict medical rater. Your only job is to output a single integer score.

Scale:
1 - Not grounded
2 - Limited extent
3 - Good extent
4 - Mostly grounded
5 - Completely grounded

Your response MUST follow this exact format: {groundedness_score: X}
where X is an integer from 1 to 5.
Do not include any explanation, introductory text, or concluding remarks."""
````

````python
medical_relevance_rater_system_message = """
You are a strict medical rater. Your only job is to output a single integer score.

Scale:
1 - Not relevant
2 - Limited extent
3 - Good extent
4 - Mostly relevant
5 - Completely relevant

Your response MUST follow this exact format: {relevance_score: X}
where X is an integer from 1 to 5.
Do not include any explanation, introductory text, or concluding remarks."""
````

````python
medical_rater_user_message_template = """
###Question
{question}

###Context
{context}

###Answer
{answer}
"""
````

````python
def generate_ground_relevance_response(user_input, response, retriever,
                                       groundedness_system_message,
                                       relevance_system_message,
                                       user_message_template,
                                       k=5, max_tokens=500, temperature=0, top_p=0.95):

    # Retrieve relevant document chunks using .invoke()
    relevant_chunks = retriever.invoke(user_input)
    if not relevant_chunks:
        context_for_rater = "No relevant context found."
    else:
        context_for_rater = "\n".join([f"Source: {d.metadata.get('source','Unknown')}\n{d.page_content}" for d in relevant_chunks])

    # Fill user message template for evaluation
    evaluation_user_message_content = user_message_template.format(
        question=user_input,
        context=context_for_rater,
        answer=response
    )

    # Groundedness evaluation
    groundedness_full_prompt = f"""{groundedness_system_message}\n\n{evaluation_user_message_content}"""
    try:
        groundedness_response = lcpp_llm(
            prompt=groundedness_full_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            repeat_penalty=1.2,
            top_k=25,
            stop=['INST'],
            echo=False
        )
        groundedness_score_text = groundedness_response["choices"][0]["text"].strip()
    except Exception as e:
        groundedness_score_text = f"Error during groundedness evaluation: {e}"

    # Relevance evaluation
    relevance_full_prompt = f"""{relevance_system_message}\n\n{evaluation_user_message_content}"""
    try:
        relevance_response = lcpp_llm(
            prompt=relevance_full_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            repeat_penalty=1.2,
            top_k=25,
            stop=['INST'],
            echo=False
        )
        relevance_score_text = relevance_response["choices"][0]["text"].strip()
    except Exception as e:
        relevance_score_text = f"Error during relevance evaluation: {e}"

    # Return the textual responses
    return groundedness_score_text, relevance_score_text
````

````python
import pandas as pd
import re

def extract_score_from_text(text):
    """
    Attempts to find a single digit 1-5 in the response text.
    """
    if not isinstance(text, str):
        return 1
    # Look for patterns like {score: 5} or just a digit 1-5 in the text
    match = re.search(r'(\b[1-5]\b)', text)
    if match:
        return int(match.group(1))
    return 1

# Re-evaluating scores for Prompt Engineering
prompt_evaluation_df = pd.DataFrame({
    "question": [question_1, question_2, question_3, question_4],
    "base_prompt_response": [
        prompt_result_df['prompt_Engineering_responses'][0],
        prompt_result_df['prompt_Engineering_responses'][1],
        prompt_result_df['prompt_Engineering_responses'][2],
        prompt_result_df['prompt_Engineering_responses'][3]
    ],
    "groundedness_score": [
        extract_score_from_text(llm_judge_prompt_ground_1),
        extract_score_from_text(llm_judge_prompt_ground_2),
        extract_score_from_text(llm_judge_prompt_ground_3),
        extract_score_from_text(llm_judge_prompt_ground_4)
    ],
    "relevance_score": [
        extract_score_from_text(llm_judge_prompt_rel_1),
        extract_score_from_text(llm_judge_prompt_rel_2),
        extract_score_from_text(llm_judge_prompt_rel_3),
        extract_score_from_text(llm_judge_prompt_rel_4)
    ]
})

print("Prompt Engineering Evaluation Results:")
display(prompt_evaluation_df)
````

````python
import pandas as pd
import re

def extract_score_from_text(text):
    """Extracts the first digit (1-5) from common LLM judging patterns."""
    if not isinstance(text, str) or not text.strip():
        return 1 # Return 1 if text is empty or no score is found
    # Handle common patterns like 'Score: 4', 'Score is 5', 'Groundedness: 4', etc.
    match = re.search(r'(?:score|groundedness|relevance)\D*([1-5])', text.lower())
    if match:
        return int(match.group(1))
    # Fallback to any standalone digit 1-5
    any_digit = re.search(r'\b([1-5])\b', text)
    if any_digit:
        return int(any_digit.group(1))
    return 1 # Return 1 if no score is explicitly found

# Populate Prompt Engineering DataFrame using extracted scores from previous judge outputs
prompt_evaluation_df = pd.DataFrame({
    "question": [question_1, question_2, question_3, question_4],
    "groundedness_score": [
        extract_score_from_text(llm_judge_prompt_ground_1),
        extract_score_from_text(llm_judge_prompt_ground_2),
        extract_score_from_text(llm_judge_prompt_ground_3),
        extract_score_from_text(llm_judge_prompt_ground_4)
    ],
    "relevance_score": [
        extract_score_from_text(llm_judge_prompt_rel_1),
        extract_score_from_text(llm_judge_prompt_rel_2),
        extract_score_from_text(llm_judge_prompt_rel_3),
        extract_score_from_text(llm_judge_prompt_rel_4)
    ]
})

# Define rag_questions using the question variables
rag_questions = [question_1, question_2, question_3, question_4]

# Construct rag_scores using the previously generated RAG evaluation results
rag_scores = [
    {
        'groundedness': extract_score_from_text(RAG_ground_1),
        'relevance': extract_score_from_text(RAG_rel_1)
    },
    {
        'groundedness': extract_score_from_text(RAG_ground_2),
        'relevance': extract_score_from_text(RAG_rel_2)
    },
    {
        'groundedness': extract_score_from_text(RAG_ground_3),
        'relevance': extract_score_from_text(RAG_rel_3)
    },
    {
        'groundedness': extract_score_from_text(RAG_ground_4),
        'relevance': extract_score_from_text(RAG_rel_4)
    }
]

# RAG scores are now available in the rag_scores list
RAG_evaluation_df = pd.DataFrame({
    "question": rag_questions,
    "groundedness_score": [s['groundedness'] for s in rag_scores],
    "relevance_score": [s['relevance'] for s in rag_scores]
})

print("-- - Base Prompt Engineering Scores ---")
display(prompt_evaluation_df)
print("\n-- - RAG Pipeline Scores ---")
display(RAG_evaluation_df)
````
